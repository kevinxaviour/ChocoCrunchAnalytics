```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('featureeng.csv',index_col=0)
df=df.replace({np.nan: None})
df.columns
```

```python
df['code'] = df['code'].astype(str).str.strip()
df.drop_duplicates(subset=['code'], inplace=True)

product_info=df[['brands','code', 'product_name']]
```

```python
import pymysql
import pandas as pd

# Example DataFrame (you already have this)
# product_info = df[['brands', 'code', 'product_name']]

# Connect to MySQL
connection = pymysql.connect(
    host='localhost',        # or your host
    user='root',
    password='12345678',
    database='guviprojects',
    charset='utf8mb4',
    cursorclass=pymysql.cursors.DictCursor
)

# Insert data
try:
    with connection.cursor() as cursor:
        sql = """
        INSERT INTO product_info (brand, product_code, product_name)
        VALUES (%s, %s, %s)
        """
        
        data = [
            (row['brands'], row['code'], row['product_name'])
            for _, row in product_info.iterrows()
        ]
        
        cursor.executemany(sql, data)
        connection.commit()
        print("✅ Data inserted successfully!")
        
except Exception as e:
    print("❌ Error:", e)
    
finally:
    connection.close()
```

```python
nutrient_info=df[['code', 'energy-kcal_value',
       'energy-kj_value', 'carbohydrates_value', 'sugars_value', 'fat_value',
       'saturated-fat_value', 'proteins_value', 'salt_value', 'sodium_value',
       'fruits-vegetables-nuts-estimate-from-ingredients_100g',
        'nutrition-score-fr','nova-group'
       ]]
```
