```python
import requests
import pandas as pd
```

```python
all_rec=[]
for page in range(1,121):
    url=f"https://world.openfoodfacts.org/api/v2/search?categories=chocolates&fields=code,product_name,brands,nutriments&page_size=100&page={page}"
    response=requests.get(url)
    if response.status_code == 200:
        data=response.json()
        all_rec.extend(data['products'])
        print(f"Successfully Fetched page:{page}")
    else:
        print(f"failed to fetch {page}:{response.status_code}")
        
    
```

```python
all_rec
```

```python
df=pd.DataFrame(all_rec)
```

```python
nutriments_expanded = pd.json_normalize(df['nutriments'])
df = df.join(nutriments_expanded)
df=df.drop('nutriments',axis=1)
columns_to_keep=['brands','code','product_name','energy-kcal_value','energy-kj_value','carbohydrates_value','sugars_value',
                'fat_value','saturated-fat_value','proteins_value','fiber_value','salt_value',
                'sodium_value','nova-group','nutrition-score-fr','fruits-vegetables-nuts-estimate-from-ingredients_100g']
df=df[columns_to_keep]
df
```

```python
df.to_csv('ChocoCrunchAnalytics.csv')
```
